```python
# Operator Overloading 


class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def sum(self, p):
        return Point((self.x + p.x), (self.y+ p.y))
    
    def print_point(self):
        return f"x is {self.x}, y is {self.y}"

```

```python
P1 = Point(3, 2)
P2 = Point(6, 3)
```

```python
P3 = P1.sum(P2)
```

```python
P3.print_point()
```

```python
# Decorators is a function that takes a function, it creates a new function in the body, then it returns that new function

def decorator(func):
    def wrapper():
        print("I am about to execute a func....")
        func()
        print("I have executed a function....")
    return wrapper


@decorator
def say_hello():
    print("Hello")


# f = decorator(say_hello)

say_hello()
```

```python
# Instance static methods 

class Emp:
    company="CAT" #class attributes 

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    
    #Instance method
    def print_info(self):
        return f"The name is {self.name} and salary is {self.salary}"
    
    @staticmethod
    def sum(a, b):
        return a+b
    
    @classmethod
    def print_company(cls):
        print(cls.company)

    def __str__(self):
        return f"{self.name}, and salary is {self.salary}"
    
    def __len__(self):
        return len(self.name)


e1 = Emp("Rahul", 25000)
e2 = Emp("Shreyas", 35000)
```

```python
e1.print_info()
```

```python
e2.sum(10, 10)
```

```python
Emp.sum(5,5)
```

```python
e2.print_company()
```
